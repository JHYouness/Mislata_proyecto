```python
import pandas as pd

# Cargar el archivo
df_graficas = pd.read_csv('Sensores_temperatura_clase_agrupados.csv')

# Lista de entity_id que queremos conservar
temp = [
    'sensor.sensor_temperatura_1_temperature',
    'sensor.sensor_temperatura_2_temperature',
    'sensor.sensor_temperatura_3_temperature',
    'sensor.sensor_temperatura_4_temperature'
]

humi = [
    'sensor.sensor_temperatura_1_humidity',
    'sensor.sensor_temperatura_2_humidity',
    'sensor.sensor_temperatura_3_humidity',
    'sensor.sensor_temperatura_4_humidity'
]

pression = [
    'sensor.sensor_temperatura_1_pressure',
    'sensor.sensor_temperatura_2_pressure',
    'sensor.sensor_temperatura_3_pressure',
    'sensor.sensor_temperatura_4_pressure'
]

# Filtrar
df_filtrado_graficas1 = df_graficas[df_graficas['entity_id'].isin(temp)]
df_filtrado_graficas2 = df_graficas[df_graficas['entity_id'].isin(humi)]
df_filtrado_graficas3 = df_graficas[df_graficas['entity_id'].isin(pression)]


# Guardar a CSV
df_filtrado_graficas1.to_csv('sensores_gráficas_temperaturas.csv', index=False)
df_filtrado_graficas2.to_csv('sensores_gráficas_humidity.csv', index=False)
df_filtrado_graficas3.to_csv('sensores_gráficas_presion.csv', index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el CSV
df = pd.read_csv("/content/sensores_gráficas.csv")  # Reemplaza con la ruta de tu archivo

# Convertir la columna 'time' a formato datetime
df["time"] = pd.to_datetime(df["time"])

plt.figure(figsize=(12, 6))
for sensor in df["entity_id"].unique():
    datos_sensor = df[df["entity_id"] == sensor]
    plt.plot(datos_sensor["time"], datos_sensor["valor_medio"], linestyle="-", label=sensor)  # Sin marker="o"

# Personalización del gráfico
plt.xlabel("Tiempo")
plt.ylabel("Temperatura (°C)")
plt.title("Comparación de Temperaturas entre Sensores (1-4)")
plt.xticks(rotation=45)  # Rotar etiquetas del eje X para mayor legibilidad
plt.legend(title="Sensores")
plt.grid(True)

# Mostrar el gráfico
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar los datos
df = pd.read_csv('/content/entrenamiento_neuronal_calefaccion.csv')  # o pd.read_excel('tus_datos.xlsx')

# Seleccionar solo columnas numéricas
df = df.select_dtypes(include='number')

# Variable específica
variable_objetivo = 'temperatura_calefaccion_y'

# Calcular correlación de esa variable con las demás
correlaciones = df.corr()[[variable_objetivo]].drop(index=variable_objetivo)

# Crear heatmap (una sola columna)
plt.figure(figsize=(6, len(correlaciones) * 0.5))
sns.heatmap(correlaciones, annot=True, cmap='coolwarm', vmin=-1, vmax=1, cbar=True)
plt.title(f"Correlación de '{variable_objetivo}' con otras variables")
plt.tight_layout()
plt.show()
```
